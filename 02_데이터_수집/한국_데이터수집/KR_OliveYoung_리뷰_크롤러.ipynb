```python
# 스크롤다운 함수
def scroll_down(driver):      
    #driver.execute_script("window.scrollBy(0,1100);")
    driver.find_element(By.TAG_NAME, 'body').send_keys(Keys.END) 
    time.sleep(1)


cnt = 10 # 수집할 상품 개수
no = 0 # 1건 수집할 때마다 1건씩 올라감
page_cnt = math.ceil(cnt / 24) 
product_urls = []

print(f"{cnt} 건을 수집하기 위해 {page_cnt} 페이지를 조회합니다.") 
print('=' *80) 

for a in range(1,page_cnt+1) : 
    print("\n") 
    print("%s 페이지 내용 수집 시작합니다 =======================" %a)
    
    scroll_down(driver)
    html = driver.page_source 
    soup = BeautifulSoup(html, 'html.parser') 
    product_list = soup.find(id='ajaxList').find_all('li','flag li_result') 

    for i in product_list:  
        try:
            product_urls.append(i.find('a')['href'])
        except:
            product_urls.append(None)     

        no += 1
        if no >= cnt:
            break         

    print(f"현재 수집된 주소 개수: {len(product_urls)}")        
    a += 1 
    try:
        page_button = driver.find_element(By.LINK_TEXT, str(a))
        page_button.click()
        time.sleep(2)
    except:
        pass

# 결과 확인
print("\n최종 수집 완료!")
#enumerate는 앞에 번호표 같이 붙여주는 함수 / 1은 1번부터 붙여줘라는 의미 (기본값 0)
for i, url in enumerate(product_urls, 1):
    print(f"{i}. {url}") #f-string / 문자열 안에 변수나 계산식을 직접 집어넣겠다! 라고 선언
```

```python
#txt 파일에 저장하기 위해 파일 open하기 
f = open(ff_name, 'a', encoding='UTF-8') 

full_data_list = []
no = 1        # 상품 1건 추출할 때마다 1씩 증가

for url in product_urls: 
    driver.execute_script(f"window.open('{url}');")
    driver.switch_to.window(driver.window_handles[-1]) #가장 최신 화면(끝)으로 스위치
    time.sleep(4) 
    html = driver.page_source 
    soup = BeautifulSoup(html, 'html.parser') 

    #리뷰버튼 클릭
    button1 = driver.find_element(By.CSS_SELECTOR, "button.GoodsDetailTabs_tab-item__tgAnU.css-167nbif")
    # 화면에 보이지 않아도 클릭하도록 강제 클릭 권장
    driver.execute_script("arguments[0].click();", button1)
    time.sleep(1) 

    # 제품명
    try : 
        name=soup.find('h3','GoodsDetailInfo_title__Vl_IP').get_text().strip()
        print("="*30)
        print(f"[ {no} ] 제품명 : {name}")
        f.write("="*30 + "\n" )
        f.write(f"[ {no} ] 제품명 : {name}\n")
    except : 
        name = '제품명이 없습니다' 
        print("="*30)
        print(f"[ {no} ] 제품명 : {name}")
        f.write("="*30 + "\n")
        f.write(f"[ {no} ] 제품명 : {name}\n")

    # 리뷰개수
    try : 
        review_cnt=soup.find('button','ReviewArea_btn-review__gZoOZ').find('span').get_text().strip().replace(",","")
        print(f"리뷰개수 : {review_cnt}")
        print("="*30)
        f.write(f"리뷰개수 : {review_cnt}\n")
        f.write("="*30 + "\n")       
    except : 
        review_cnt = '0' 
        print(f"리뷰개수 : {review_cnt}")
        print("="*30)
        f.write(f"리뷰개수 : {review_cnt}\n")
        f.write("="*30 + "\n")

    # 목표 수집 리뷰 개수 설정
    target_count = 100

    # 실행할 자바스크립트
    review = """
    const TARGET_COUNT = arguments[0];
    const done = arguments[arguments.length - 1];

    (async function() {
        console.log("🚀 Selenium 수집 프로세스 시작...");

        try {
            // [1] 정렬 클릭 로직
            const host1 = document.querySelector('oy-review-review-in-product');
            const sortHost = host1?.shadowRoot
                ?.querySelector('oy-review-review-list')?.shadowRoot
                ?.querySelector('oy-review-review-filter')?.shadowRoot
                ?.querySelector('oy-review-review-sort');
            const sortRoot = sortHost?.shadowRoot;

            if (sortRoot) {
                const buttons = sortRoot.querySelectorAll('button.pc-sort-button');
                const targetButton = [...buttons].find(btn => btn.innerText.includes("평점 높은순"));
                if (targetButton) {
                    targetButton.click();
                    await new Promise(res => setTimeout(res, 2500)); // 정렬 후 로딩 대기
                }
            }
        } catch (e) { console.error("정렬 클릭 에러:", e); }

        // [2] 데이터 수집 로직
        const results = [];
        const seenKeys = new Set();
        let attempts = 0;

        while (results.length < TARGET_COUNT && attempts < 50) {
            const host1 = document.querySelector('oy-review-review-in-product');
            const root1 = host1?.shadowRoot;
            
            // Shadow DOM 내부 리스트 영역 탐색
            const listComponent = root1?.querySelector('oy-review-review-list')?.shadowRoot || 
                                root1?.querySelector('oy-review-product-review-provider')?.shadowRoot
                                    ?.querySelector('oy-review-review-list-provider')?.shadowRoot
                                    ?.querySelector('oy-review-review-list')?.shadowRoot;

            if (!listComponent) {
                window.scrollBy(0, 700);
                await new Promise(res => setTimeout(res, 1500));
                attempts++;
                continue;
            }

            const items = listComponent.querySelectorAll('oy-review-review-item');
            for (let item of items) {
                if (results.length >= TARGET_COUNT) break;

                const itemRoot = item.shadowRoot;
                if (!itemRoot) continue;

                // 별점 계산 (fill 속성 기반)
                const stars = itemRoot.querySelectorAll('oy-review-star-icon');
                let score = 0;
                stars.forEach(star => {
                    const path = star.shadowRoot?.querySelector('path');
                    const fill = path?.getAttribute('fill');
                    if (fill && fill !== "none" && fill !== "transparent") score++;
                });

                // 유저 정보
                const userHost = itemRoot.querySelector('oy-review-review-user');
                const userRoot = userHost ? userHost.shadowRoot : null;
                const userName = userHost?.shadowRoot.querySelector('.name')?.innerText.trim() || "알 수 없음";

                // 날짜
                const date = itemRoot.querySelector('.date')?.innerText.trim() || "";
                
                // 피부정보
                const skinTags = userRoot ? Array.from(userRoot.querySelectorAll('.skin-type')) : [];
                const skinInfo = skinTags.map(tag => tag.innerText.trim()).join(', ') || "정보 없음";

                
                // 리뷰 본문 (더보기 클릭 포함)
                const contentHost = itemRoot.querySelector('oy-review-review-content')?.shadowRoot;
                contentHost?.querySelector('.more-button')?.click(); 
                const content = contentHost?.querySelector('.content')?.innerText.trim() || "";

                const key = userName + date + content.slice(0, 20);
                if (content && !seenKeys.has(key)) {
                    seenKeys.add(key);
                    results.push({
                        user: userName,
                        skin: skinInfo,
                        rating: score,
                        date: date,
                        content: content
                    });
                }
            }
            
            if (results.length < TARGET_COUNT) {
                window.scrollBy(0, 1000);
                await new Promise(res => setTimeout(res, 2000));
            }
            attempts++;
        }

        done(results); // 파이썬으로 최종 리스트 전달
    })();
    """

    리뷰_데이터 = driver.execute_async_script(review , target_count)

    # 수집된 리뷰 루프 안에서 제품 정보를 매칭시켜 저장
    for i, data in enumerate(리뷰_데이터, 1):
        # 행(Row) 단위로 딕셔너리 생성
        row_dict = {
            "순서" : no,
            "상품명": name,
            "리뷰수": review_cnt,
            "작성자": data['user'],
            "피부정보": data['skin'],
            "내용": data['content'],
            "날짜": data['date'],
            "별점": data['rating']    }
        
        # 엑셀용 통합 리스트에 추가
        full_data_list.append(row_dict)

        # 텍스트 파일 기록
        print(f"[{i}/{target_count}] 작성자: {data['user']}\n피부정보: {data['skin']}\n내용: {data['content']}\n날짜: {data['date']}\n별점: {data['rating']}")
        print("-" * 30)
        

        f.write(f"작성자: {data['user']}\n피부정보: {data['skin']}\n내용: {data['content']}\n날짜: {data['date']}\n별점: {data['rating']}\n")
        f.write("-" * 30 + "\n")

    no += 1
    driver.close()
    driver.switch_to.window(driver.window_handles[0]) #처음(가장 오래된 것) 화면으로 돌아가라

    if no > cnt:
        break


f.close( )
```

```python
df = pd.DataFrame(full_data_list)
df.to_excel(fx_name, index=False)
print(f"엑셀 저장 완료 : {fx_name}")
```
